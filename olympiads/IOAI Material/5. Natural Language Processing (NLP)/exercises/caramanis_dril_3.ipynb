```python
from transformers import AutoModel, AutoTokenizer, AutoModelForCausalLM, AutoModelForSeq2SeqLM
import numpy as np
import torch
```

```python
model = AutoModelForCausalLM.from_pretrained("distilgpt2")
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
```

```python
query = "Write one sentence about space exploration."
```

```python
tokenized = tokenizer(query, return_tensors="pt")
generated_ids = model.generate(
    input_ids=tokenized["input_ids"], 
    attention_mask=tokenized["attention_mask"],
    do_sample=True,
    temperature=.8,
    max_new_tokens=30,
    top_p=.95,
    pad_token_id=tokenizer.eos_token_id
)
print(generated_ids)
print(tokenizer.decode(generated_ids[0], skip_special_tokens=True))
```

distilgpt2 is decoder-only. It does not answer instructions reliably; it continues the prompt by predicting next tokens.
