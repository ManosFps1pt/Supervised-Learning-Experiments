## Tokenization note

For this warm-up, simple word-level tokenization is intentional.

- lowercase each sentence,
- split into words,
- use the same rule for every sentence.

This is enough for a first bigram language-model exercise. `word2vec` is not a tokenizer; it is a later representation method.

```python
X = [ "i like pizza",
 "i like pasta",
 "you like pizza",
 "you cook pasta",
 "we cook dinner",
 "we like dinner",
 "they cook rice",
 "they like rice"
 ]
```

```python
def tokenize(x):
    return [i.lower().split(" ") for i in x]

print(tokenize(X))
```
