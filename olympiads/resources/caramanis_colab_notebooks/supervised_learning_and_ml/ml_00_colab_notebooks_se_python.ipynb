# Βασικές Ιδέες της Μηχανικής Μάθησης

Εδώ θα βρείτε μία σειρά από Colab Notebooks που αναπτύσσουν βασικές ιδέες, αλγορίθμους και μεθόδους από την Μηχανική Μάθηση (Machine Learning).

Τα Notebooks είναι σχεδιασμένα να τα δείτε με την σειρά αρίθμησης, συνοδευόμενα από τα βιντεοσκοπημένα μαθήματα που μπαίνουν σε λεπτομερή συζήτηση για κάθε αλγοριθμική ιδέα, και εντολή. Όταν πρωτοεισάγουμε κάποια καινούργια ιδέα ή εντολή, εξηγούμε την χρήση της λεπτομερώς στο Notebook. Δεν μπαίνουμε στο ίδιο βάθος συζήτησης και εξήγησης οταν εμφανίζεται μιά ιδέα/εντολή για 2η ή 3η φορά.

Τα Notebooks είναι όλα ``read-only``, οπότε πρέπει να αποθηκεύσετε ένα αντίγραφο στο δικό σας Google Drive.

**Χρήση GPU**: για την εκπαίδευση νευρωνικών δικτύων, θα βοηθήσει πολύ η χρήση του GPU. Η Colab προσφέρει δωρεάν GPU εάν είναι διαθέσιμα την στιγμή που τα ζητάμε (``Runtime`` --> ``Change runtime type`` --> ``T4 GPU``). Είναι δυνατόν να αγοραστούν credits που μπορούν να διευκολύνουν τα πειράματά μας (τους υπολογισμούς) διοτί με credits, τα GPU της Colab μας είναι πάντα διαθέσιμα (``Runtime``-->``View Resources``-->``Learn more``-->``Pay as you go``).
