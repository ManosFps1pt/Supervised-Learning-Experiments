## Υλοποίηση Bag of Words Search

Υλοποιούμε ένα απλό σύστημα αναζήτησης Bag-of-Words. Αυτό θα χρησιμεύσει ως σημείο σύγκρισης με πιο εξελιγμένες μεθόδους ανάκτησης: TF-IDF σε αυτό το notebook, και σημασιολογική αναζήτηση στο επόμενο.


### Core Classes

Πρώτα, υλοποιούμε ένα βασικό σύστημα information retrieval χρησιμοποιώντας το μοντέλο Bag of Words (BoW). Η κλάση `BagOfWordsSearch` παρέχει έναν απλό αλλά αποτελεσματικό τρόπο αναζήτησης σε έγγραφα με βάση επικάλυψη λέξεων.

### Key Components

#### Initialization
- Δημιουργεί ένα άδειο αποθετήριο εγγράφων
- Προαιρετική υποστήριξη stemming χρησιμοποιώντας Porter Stemmer (μειώνει τις λέξεις στη ρίζα τους)

#### Document Processing
- **Text Preprocessing** (`_preprocess_text`):
  - Μετατρέπει το κείμενο σε πεζά γράμματα
  - Αφαιρεί χαρακτήρες που δεν αντιστοιχούν σε γράμματα ή αριθμούς
  - Εφαρμόζει προαιρετικά stemming (π.χ., "running" → "run")

#### Search Functionality
- **Search Method**:
  - Παίρνει ένα query string και προαιρετική παράμετρο `top_k`
  - Υπολογίζει τη σχετικότητα εγγράφων χρησιμοποιώντας την επικάλυψη: set intersection
  - Score = αριθμός επικαλυπτόμενων λέξεων μεταξύ query και εγγράφου
  - Επιστρέφει έγγραφα καταταγμένα κατά relevance score



````python
class BagOfWordsSearch:
    def __init__(self, use_stemming: bool = False):
        """Initialize the Bag of Words search system."""
        self.documents: Dict[str, str] = {}
        self.use_stemming = use_stemming
        if use_stemming:
            self.stemmer = PorterStemmer()

    def add_document(self, doc_id: str, text: str) -> None:
        """Add a document to the search index."""
        self.documents[doc_id] = text

    def _preprocess_text(self, text: str) -> List[str]:
        """Preprocess text by tokenizing and optionally stemming."""
        tokens = word_tokenize(text.lower())
        tokens = [re.sub(r'[^a-z0-9]', '', token) for token in tokens]
        tokens = [token for token in tokens if token]

        if self.use_stemming:
            tokens = [self.stemmer.stem(token) for token in tokens]

        return tokens

    def search(self, query: str, top_k: Optional[int] = None) -> List[tuple]:
        """Search documents using bag of words matching."""
        query_tokens = set(self._preprocess_text(query))
        results = []

        for doc_id, text in self.documents.items():
            doc_tokens = set(self._preprocess_text(text))
            score = len(query_tokens & doc_tokens) # '&' is the intersection operator
            results.append((doc_id, score))

        results.sort(key=lambda x: x[1], reverse=True)

        if top_k is not None:
            results = results[:top_k]

        return results
````

## Sparse Search: TF-IDF

Το TF-IDF μετρά πόσο σημαντική είναι μια λέξη σε ένα έγγραφο σε μια συλλογή. Συνδυάζει δύο παράγοντες:
1. Πόσο συχνά εμφανίζεται η λέξη στο έγγραφο (TF)
2. Πόσο μοναδική είναι η λέξη σε όλα τα έγγραφα (IDF)

### Term Frequency (TF)
- Μετρά πόσο συχνά εμφανίζεται μια λέξη σε ένα έγγραφο
- Περισσότερες εμφανίσεις → πιο σημαντική
```python
tf = (αριθμός φορών που εμφανίζεται ο όρος στο έγγραφο) / (συνολικές λέξεις στο έγγραφο)
```

### Inverse Document Frequency (IDF)
- Μετρά πόσο μοναδική ή σπάνια είναι μια λέξη σε όλα τα έγγραφα
- Οι σπάνιες λέξεις παίρνουν υψηλότερα scores, οι κοινές λέξεις παίρνουν χαμηλότερα scores
```python
idf = log(συνολικός αριθμός εγγράφων / αριθμός εγγράφων που περιέχουν τον όρο)
```

### Συνδυασμός
- TF-IDF = TF × IDF
- Υψηλό score σημαίνει:
  - Η λέξη εμφανίζεται συχνά σε αυτό το έγγραφο (υψηλό TF)
  - Η λέξη εμφανίζεται σε λίγα άλλα έγγραφα (υψηλό IDF)

### Παράδειγμα
Σκεφτείτε τη λέξη "neural" vs "transformer" σε ερευνητικά papers για νευρωνικά δίκτυα:
- "neural":
  - Υψηλό TF (εμφανίζεται συχνά)
  - Χαμηλό IDF (εμφανίζεται σε κάθε έγγραφο)
  - Χαμηλό TF-IDF score
- "transformer":
  - Μέτριο TF (μπορεί να εμφανίζεται αρκετές φορές)
  - Υψηλό IDF (εμφανίζεται σε λιγότερα έγγραφα)
  - Υψηλότερο TF-IDF score

Αυτός είναι ο λόγος που το TF-IDF είναι καλύτερο από την απλή μέτρηση λέξεων (επικάλυψη) - χειρίζεται αυτόματα τη σημαντικότητα των λέξεων με βάση τη μοναδικότητά τους στο συνολικό σώμα κειμένων (text corpus).

### TF-IDF: βήμα προς βήμα

Γιατί πετυχαίνει το TF-IDF; Ας υπολογίσουμε ρητά το TF-IDF score του.


````python
import re
import math
from collections import Counter
from nltk.tokenize import word_tokenize

def preprocess_text(text):
    """Preprocess text: lowercase, tokenize, remove non-alphanumeric."""
    tokens = word_tokenize(text.lower())
    tokens = [re.sub(r'[^a-z0-9]', '', token) for token in tokens]
    tokens = [token for token in tokens if token]
    return tokens

# Select documents to analyze
doc_correct = toy_corpus[15]  # The correct document (Doc 16 in 1-based)
doc_high_overlap_1 = toy_corpus[10]  # Doc 11 - high word overlap
doc_high_overlap_2 = toy_corpus[18]  # Doc 19 - high word overlap

# Preprocess query and documents
query_tokens = preprocess_text(toy_query)
doc_correct_tokens = preprocess_text(doc_correct)
doc_high1_tokens = preprocess_text(doc_high_overlap_1)
doc_high2_tokens = preprocess_text(doc_high_overlap_2)

# Preprocess all documents to compute IDF
all_doc_tokens = [preprocess_text(doc) for doc in toy_corpus]
total_docs = len(toy_corpus)

print("=" * 80)
print("TF-IDF Calculation - Step by Step")
print("=" * 80)
print(f"\nQuery: '{toy_query}'")
print(f"Query tokens: {query_tokens}")
print(f"\nDocuments to analyze:")
print(f"  Doc 16 (correct): {doc_correct[:80]}...")
print(f"  Doc 11 (high overlap): {doc_high_overlap_1[:80]}...")
print(f"  Doc 19 (high overlap): {doc_high_overlap_2[:80]}...")

# Step 1: Calculate IDF for each query term across all documents
print("\n" + "=" * 80)
print("Step 1: Calculate IDF (Inverse Document Frequency)")
print("=" * 80)
print("\nIDF = log(total_documents / documents_containing_term)")
print(f"Total documents: {total_docs}\n")

idf_scores = {}
for term in query_tokens:
    # Count how many documents contain this term
    docs_with_term = sum(1 for doc_tokens in all_doc_tokens if term in doc_tokens)
    if docs_with_term > 0:
        idf = math.log(total_docs / docs_with_term)
        idf_scores[term] = idf
        print(f"  '{term}': appears in {docs_with_term}/{total_docs} docs → IDF = log({total_docs}/{docs_with_term}) = {idf:.4f}")
    else:
        idf_scores[term] = 0
        print(f"  '{term}': appears in 0/{total_docs} docs → IDF = 0 (term not in corpus)")

# Step 2: Calculate TF for each document and query term
print("\n" + "=" * 80)
print("Step 2: Calculate TF (Term Frequency) for each document")
print("=" * 80)
print("\nTF = (count of term in document) / (total terms in document)\n")

def calculate_tf(doc_tokens, term):
    """Calculate term frequency in document."""
    term_count = doc_tokens.count(term)
    total_terms = len(doc_tokens)
    if total_terms > 0:
        return term_count / total_terms, term_count, total_terms
    return 0, 0, 0

# Calculate TF for each document
print("Doc 16 (correct):")
doc_correct_tf = {}
for term in query_tokens:
    tf, count, total = calculate_tf(doc_correct_tokens, term)
    doc_correct_tf[term] = tf
    print(f"  '{term}': appears {count} times in {total} total terms → TF = {count}/{total} = {tf:.4f}")

print("\nDoc 11 (high overlap):")
doc_high1_tf = {}
for term in query_tokens:
    tf, count, total = calculate_tf(doc_high1_tokens, term)
    doc_high1_tf[term] = tf
    print(f"  '{term}': appears {count} times in {total} total terms → TF = {count}/{total} = {tf:.4f}")

print("\nDoc 19 (high overlap):")
doc_high2_tf = {}
for term in query_tokens:
    tf, count, total = calculate_tf(doc_high2_tokens, term)
    doc_high2_tf[term] = tf
    print(f"  '{term}': appears {count} times in {total} total terms → TF = {count}/{total} = {tf:.4f}")

# Step 3: Calculate TF-IDF for each document
print("\n" + "=" * 80)
print("Step 3: Calculate TF-IDF = TF × IDF for each query term")
print("=" * 80)

def calculate_tfidf(tf_dict, idf_dict):
    """Calculate TF-IDF scores for all query terms."""
    tfidf_dict = {}
    total_tfidf = 0
    for term in query_tokens:
        tf = tf_dict.get(term, 0)
        idf = idf_dict.get(term, 0)
        tfidf = tf * idf
        tfidf_dict[term] = tfidf
        total_tfidf += tfidf
    return tfidf_dict, total_tfidf

doc_correct_tfidf, doc_correct_score = calculate_tfidf(doc_correct_tf, idf_scores)
doc_high1_tfidf, doc_high1_score = calculate_tfidf(doc_high1_tf, idf_scores)
doc_high2_tfidf, doc_high2_score = calculate_tfidf(doc_high2_tf, idf_scores)

print("\nDoc 16 (correct):")
for term in query_tokens:
    tf = doc_correct_tf[term]
    idf = idf_scores.get(term, 0)
    tfidf = doc_correct_tfidf[term]
    print(f"  '{term}': TF({tf:.4f}) × IDF({idf:.4f}) = {tfidf:.4f}")
print(f"  Total TF-IDF score: {doc_correct_score:.4f}")

print("\nDoc 11 (high overlap):")
for term in query_tokens:
    tf = doc_high1_tf[term]
    idf = idf_scores.get(term, 0)
    tfidf = doc_high1_tfidf[term]
    print(f"  '{term}': TF({tf:.4f}) × IDF({idf:.4f}) = {tfidf:.4f}")
print(f"  Total TF-IDF score: {doc_high1_score:.4f}")

print("\nDoc 19 (high overlap):")
for term in query_tokens:
    tf = doc_high2_tf[term]
    idf = idf_scores.get(term, 0)
    tfidf = doc_high2_tfidf[term]
    print(f"  '{term}': TF({tf:.4f}) × IDF({idf:.4f}) = {tfidf:.4f}")
print(f"  Total TF-IDF score: {doc_high2_score:.4f}")

# Step 4: Summary and comparison
print("\n" + "=" * 80)
print("Step 4: Summary and Key Insights")
print("=" * 80)
print("\nFinal TF-IDF Scores (sum of TF-IDF for all query terms):")
print(f"  Doc 16 (correct):      {doc_correct_score:.4f}")
print(f"  Doc 11 (high overlap): {doc_high1_score:.4f}")
print(f"  Doc 19 (high overlap): {doc_high2_score:.4f}")

print("\n🔍 Key Observations:")
if 'cimetidine' in idf_scores:
    print(f"  1. 'cimetidine' has the highest IDF ({idf_scores.get('cimetidine', 0):.4f}) - it's the rarest term!")
print(f"  2. Doc 16 contains 'cimetidine', giving it a significant boost despite lower TF for common terms")
print(f"  3. Docs 11 and 19 have high TF for 'metformin' and 'interaction', but these terms have lower IDF")
print(f"  4. The rare term 'cimetidine' outweighs the common terms in TF-IDF scoring")
print("\n✅ This explains why TF-IDF ranks Doc 16 higher than BoW!")
````

### Ποσοτική Αξιολόγηση: NDCG & RECALL

Τέλος, ας αξιολογήσουμε ποσοτικά την αναζήτηση Bag-of-Words χρησιμοποιώντας το NDCG και το Recall.


## Υπολογισμός Μετρικών Αξιολόγησης

**Τι κάνει ένας καλός αλγόριθμος ανάκτησης;** Διαισθητικά, πρέπει να κατατάσσει τα πιο σχετικά έγγραφα ψηλότερα από τα λιγότερο σχετικά έγγραφα. Υπάρχουν πολλοί τρόποι με τους οποίους αυτό μπορεί να ποσοτικοποιηθεί.

Θα υπολογίσουμε το Normalized Discounted Cumulative Gain ([nDCG@K](https://en.wikipedia.org/wiki/Discounted_cumulative_gain)) και το [Recall@K](https://en.wikipedia.org/wiki/Precision_and_recall).

### Συνοπτικά: nDCG vs Recall

**nDCG**: Ένα καλό αποτέλεσμα ``nDCG@K`` μας λέει ότι ο αλγόριθμος ανάκτησης κατατάσσει μερικά από τα πιο σχετικά έγγραφα πολύ ψηλά (αλλά μπορεί κάποια σχετικά έγγραφα να μην εμφανίζονται καν στα πρώτα Κ).

**Recall**: Ένα καλό αποτέλεσμα ``Recall@K`` μας λέει ότι μεγάλο ποσοστό των σχετικών εγγράφων κατατάσσονται στα πρώτα Κ αποτελέσματα (αλλά μπορεί να μην βρίσκονται πολύ ψηλά στη λίστα).


## NDCG και Recall: οι λεπτομέρειες των υπολογισμών

### 1. Normalized Discounted Cumulative Gain (nDCG@K)
Το **nDCG@K** είναι μια μετρική αξιολόγησης μιας κατάταξης, που ποσοτικοποιεί πόσο καλά μια μέθοδος αναζήτησης ή ένα ranking μοντέλο ταξινομεί σχετικά έγγραφα στα πρώτα $K$ αποτελέσματα. Λαμβάνει υπόψη και τα δύο:
- **Relevance** των ανακτηθέντων εγγράφων.
- **Position** (τα ψηλότερα στην κατάταξη σχετικά έγγραφα είναι πιο πολύτιμα).

Το nDCG υπολογίζεται ως:

$$
\text{nDCG@K} = \frac{\text{DCG@K}}{\text{IDCG@K}}
$$

όπου:

1. **Discounted Cumulative Gain (DCG@K)**:

$$
\text{DCG@K} = \sum_{i=1}^{K} \frac{\text{rel}_i}{\log_2(i + 1)}
$$

- $ \text{rel}_i $ = relevance score του εγγράφου στη θέση $ i $.
- $ \log_2(i + 1) $ μειώνει τη σημασία των χαμηλότερα στην κατάταξη σχετικών εγγράφων.

2. **Ideal DCG (IDCG@K)**:
- Το **καλύτερο δυνατό DCG**: έγγραφα ταξινομημένα σε ιδανική/τέλεια σειρά: $\text{rel}_1 > \text{rel}_2 > \text{rel}_3$ κ.ο.κ.
- Υπολογίζεται με τον ίδιο τρόπο με το DCG αλλά χρησιμοποιώντας την **ιδανική κατάταξη**.

Έτσι, το **nDCG@K είναι πάντα μεταξύ 0 και 1**, όπου:
- **1.0** σημαίνει τέλεια κατάταξη.
- **Χαμηλότερες τιμές** υποδηλώνουν υποβέλτιστη κατάταξη.

#### Παράδειγμα Υπολογισμού
Ας φανταστούμε ότι κατατάσσουμε **3 έγγραφα** όπου τα relevance scores τους (ground truth) είναι **[3, 2, 1]**.

Εάν το μοντέλο τα κατατάσσει ως **[2, 3, 1]**, υπολογίζουμε:

$$
\text{DCG@3} = \frac{2}{\log_2(1+1)} + \frac{3}{\log_2(2+1)} + \frac{1}{\log_2(3+1)}
$$

$$
= 2 + \frac{3}{1.585} + \frac{1}{2} \approx 3.892
$$

Για το **IDCG@3** (ιδανική κατάταξη [3,2,1]):

$$
\text{IDCG@3} = \frac{3}{\log_2(1+1)} + \frac{2}{\log_2(2+1)} + \frac{1}{\log_2(3+1)}
$$

$$
= 3 + \frac{2}{1.585} + \frac{1}{2} \approx 4.892
$$

$$
\text{nDCG@3} = \frac{3.892}{4.892} \approx 0.796
$$

Αυτό σημαίνει ότι η κατάταξή μας είναι **79.6% τόσο καλή** όσο μια ιδανική κατάταξη.

---

### 2. Recall@K
Το **Recall@K** μετρά πόσα σχετικά έγγραφα ανακτώνται στα πρώτα $K$ αποτελέσματα, σε σύγκριση με όλα τα σχετικά έγγραφα στο dataset.

#### Formula
$$
\text{Recall@K} = \frac{\text{Αριθμός σχετικών εγγράφων στα πρώτα K}}{\text{Συνολικός αριθμός σχετικών εγγράφων στο corpus}}
$$

- Εάν ένα dataset έχει **10 σχετικά έγγραφα** και το σύστημα ανακτά **4 από αυτά στα top 5**, τότε:
$$
  \text{Recall@5} = \frac{4}{10} = 0.4
$$
  που σημαίνει ότι το σύστημα ανακάλυψε **40% όλων των σχετικών εγγράφων** στα top 5 αποτελέσματα.

#### Ερμηνεία
- **Υψηλό Recall@K** → Το μοντέλο βρίσκει περισσότερα σχετικά έγγραφα.
- **Χαμηλό Recall@K** → Πολλά σχετικά έγγραφα λείπουν από τα top αποτελέσματα.

#### Σύγκριση με Precision@K
Το Precision@K είναι μια άλλη μετρική αξιολόγησης που επίσης χρησιμοποιείται συχνά, και ποσοτικοποιεί πόσα από τα πρώτα Κ άρθρα στην κατάταξη είναι σχετικά. Οπότε για το παράδειγμά μας με 10 σχετικά έγγραφα όπου το σύστημα ανακτά 4 από αυτά στα πρώτα 5, τότε
$$
  \text{Precision@5} = \frac{4}{5} = 0.4
$$
Εάν υπάρχουν 2 σχετικά άρθρα και το σύστημα ανακτά και τα 2 στα πρώτα πέντε, τότε έχουμε
$$
  \text{Recall@5} = \frac{2}{2} = 1
$$
$$
  \text{Precision@5} = \frac{2}{5} = 0.4
$$


---

### Πότε να Χρησιμοποιήσουμε nDCG@K vs Recall@K
- Το **nDCG@K** είναι χρήσιμο όταν **η σειρά κατάταξης έχει σημασία**, όπως σε search engines.
- Το **Recall@K** είναι καλύτερο όταν σας ενδιαφέρει να **ανακτήσετε όλα τα σχετικά docs** (π.χ., αναζήτηση νομικών εγγράφων, ή αν χτίζετε ένα RAG).


````python
def calculate_metrics(qrels, results, k_values):
    """Calculate NDCG and Recall metrics for search results."""
    ndcg = {f"NDCG@{k}": 0.0 for k in k_values}
    recall = {f"Recall@{k}": 0.0 for k in k_values}

    for query_id in qrels:
        if query_id not in results:
            continue

        relevant_docs = set(qrels[query_id].keys())
        retrieved_docs = list(results[query_id].keys())

        # Calculate NDCG
        for k in k_values:
            dcg = 0
            idcg = sum((1 / math.log2(i + 2) for i in range(min(k, len(relevant_docs)))))

            for i, doc_id in enumerate(retrieved_docs[:k]):
                if doc_id in relevant_docs:
                    dcg += 1 / math.log2(i + 2)

            if idcg > 0:
                ndcg[f"NDCG@{k}"] += dcg / idcg

        # Calculate Recall
        for k in k_values:
            retrieved_relevant = len(set(retrieved_docs[:k]) & relevant_docs)
            if len(relevant_docs) > 0:
                recall[f"Recall@{k}"] += retrieved_relevant / len(relevant_docs)

    # Average the metrics
    num_queries = len(qrels)
    for k in k_values:
        ndcg[f"NDCG@{k}"] /= num_queries
        recall[f"Recall@{k}"] /= num_queries

    return ndcg, recall
````

## Πολυπλοκότητα & Search Indexing

**Η διαδικασία αναζήτησης δεν ήταν αποδοτική** Χρειάστηκε πολύς χρόνος μόνο για να επεξεργαστούμε ``sample_size=10`` queries. Τι θα γινόταν αν θέλαμε να επεξεργαστούμε όλα;

### Το Πρόβλημα: Γραμμική Αναζήτηση (Linear Search)
Χωρίς ευρετήριο (index), η αναζήτηση σε έγγραφα είναι σαν να ψάχνει κανείς σε κάθε σελίδα κάθε βιβλίου σε μια βιβλιοθήκη για να βρούμε αναφορές μιας λέξης. Για κάθε query:
- Πρέπει να κοιτάξουμε κάθε έγγραφο
- Για κάθε έγγραφο, πρέπει να σαρώσουμε όλες τις λέξεις του
- Πολυπλοκότητα χρόνου: O(N × M) όπου N = αριθμός εγγράφων, M = μέσο μήκος εγγράφου

### Η Λύση: Inverted Index
Ένα inverted index είναι σαν ένα ευρετήριο βιβλίου αλλά για ολόκληρη τη συλλογή εγγράφων. Αντί να οργανώνουμε πρώτα κατά έγγραφα, οργανώνουμε κατά λέξεις:

```python
# Χωρίς index (αρχική προσέγγιση):
documents = {
    'doc1': 'the cat sat',
    'doc2': 'the dog ran',
    'doc3': 'cat and dog'
}

# Με inverted index:
index = {
    'the': {'doc1', 'doc2'},
    'cat': {'doc1', 'doc3'},
    'dog': {'doc2', 'doc3'},
    'sat': {'doc1'},
    'ran': {'doc2'},
    'and': {'doc3'}
}
```

### Γιατί είναι Πιο Γρήγορο

**Αρχική Αναζήτηση (Linear)**
Για query "cat dog":
1. Κοιτάμε doc1: σαρώνουμε όλες τις λέξεις → βρίσκουμε 'cat'
2. Κοιτάμε doc2: σαρώνουμε όλες τις λέξεις → βρίσκουμε 'dog'
3. Κοιτάμε doc3: σαρώνουμε όλες τις λέξεις → βρίσκουμε και τα δύο
- Πρέπει να εξετάσουμε ΚΑΘΕ λέξη σε ΚΑΘΕ έγγραφο

**Indexed Search**
Για query "cat dog":
1. Αναζητούμε 'cat' → παίρνουμε αμέσως {doc1, doc3}
2. Αναζητούμε 'dog' → παίρνουμε αμέσως {doc2, doc3}
3. Συνδυάζουμε αποτελέσματα → το doc3 έχει και τις δύο λέξεις
- Εξετάζουμε μόνο τα σχετικά έγγραφα
- Άμεση πρόσβαση σε θέσεις λέξεων

### Σύγκριση Απόδοσης
- **Linear Search**: O(N × M) πράξεις
  - N = αριθμός εγγράφων
  - M = μέσο μήκος εγγράφου
  
- **Indexed Search**: O(k × |posting_list|) πράξεις
  - k = αριθμός όρων query
  - |posting_list| = μέσος αριθμός εγγράφων ανά λέξη
  - Συνήθως |posting_list| << N × M

### Συμβιβάσεις
- **Χώρος**: Το index απαιτεί πρόσθετη αποθήκευση
- **Χρόνος Προεπεξεργασίας**: Πρέπει να χτιστεί το ευρετήριο πριν την αναζήτηση
- **Κόστος Ενημέρωσης**: Η προσθήκη/τροποποίηση εγγράφων απαιτεί ενημερώσεις στο ευρετήριο
- **Όφελος**: Δραματικά ταχύτερος χρόνος αναζήτησης

**Απλό Παράδειγμα**
Με 10.000 έγγραφα που έχουν κατά μέσο όρο 1.000 λέξεις το καθένα:
- Linear search: Πρέπει να εξετάσουμε 10.000.000 λέξεις ανά query
- Indexed search: Μπορεί να εξετάσουμε μόνο μερικές εκατοντάδες document IDs ανά λέξη query

Αυτός είναι ο λόγος που η τροποποιημένη υλοποίηση Bag of Words μας με index μπορεί να χειριστεί αποτελεσματικά το πλήρες dataset, και ο λόγος που αρχικά περιοριστήκαμε σε 10 queries.

````python
class BagOfWordsSearch2:
    def __init__(self, use_stemming: bool = False):
        """Initialize the Bag of Words search system with an inverted index."""
        self.documents = {}  # Document storage
        self.index = {}     # Inverted index: word -> set of doc_ids
        self.use_stemming = use_stemming
        if use_stemming:
            self.stemmer = PorterStemmer()

    def add_document(self, doc_id: str, text: str) -> None:
        """Add a document to the search index."""
        # Store the original document
        self.documents[doc_id] = text

        # Process the document
        tokens = self._preprocess_text(text)

        # Add each token to the inverted index
        for token in tokens:
            if token not in self.index:
                self.index[token] = set()
            self.index[token].add(doc_id)

    def _preprocess_text(self, text: str) -> List[str]:
        """Preprocess text by tokenizing and optionally stemming."""
        tokens = word_tokenize(text.lower())
        tokens = [re.sub(r'[^a-z0-9]', '', token) for token in tokens]
        tokens = [token for token in tokens if token]

        if self.use_stemming:
            tokens = [self.stemmer.stem(token) for token in tokens]

        return tokens

    def search(self, query: str, top_k: Optional[int] = None) -> List[tuple]:
        """Search documents using bag of words matching with inverted index."""
        query_tokens = self._preprocess_text(query)
        doc_scores = {}  # doc_id -> score

        # For each query token, find matching documents using the index
        for token in query_tokens:
            if token in self.index:
                for doc_id in self.index[token]:
                    doc_scores[doc_id] = doc_scores.get(doc_id, 0) + 1

        # Convert to list and sort by score
        results = [(doc_id, score) for doc_id, score in doc_scores.items()]
        results.sort(key=lambda x: x[1], reverse=True)

        if top_k is not None:
            results = results[:top_k]

        return results

# Now let's use it on the full dataset
print("Initializing search engine...")
bow_search = BagOfWordsSearch2(use_stemming=True)

# Add all documents to the index
print("\nBuilding index...")
for doc_id, doc in tqdm(corpus.items(), desc="Indexing documents"):
    full_text = f"{doc['title']} {doc['text']}"
    bow_search.add_document(doc_id, full_text)

# Run all queries
print("\nRunning searches...")
bow_results = {}
for query_id, query_text in tqdm(queries.items(), desc="Processing queries"):
    results = bow_search.search(query_text, top_k=max(k_values))
    bow_results[query_id] = results

# Calculate metrics using the same code as before
formatted_results = {
    query_id: {doc_id: score for doc_id, score in results}
    for query_id, results in bow_results.items()
}

# Calculate metrics
ndcg, recall = calculate_metrics(qrels, formatted_results, k_values)

# Display results
print("\nBag of Words Search Metrics (Full Dataset):")
for k in k_values:
    print(f"NDCG@{k}: {ndcg[f'NDCG@{k}']:.4f}")
    print(f"Recall@{k}: {recall[f'Recall@{k}']:.4f}")
````

### Ανάκτηση

Θα ανακτήσουμε τώρα έγγραφα με την Pyserini. Έχουμε ήδη δει πώς η Pyserini χτίζει το ευρετήριο (inverted index). Η Pyserini διαθέτει επίσης βιβλιοθήκες για την ανάκτηση. Ενώ στο πρώτο μέρος του notebook υλοποιήσαμε μόνοι μας το TF-IDF, εδώ χρησιμοποιούμε τη βιβλιοθήκη ``LuceneSearcher`` της Pyserini.

Το `LuceneSearcher` χρησιμοποιεί BM25, μια μέθοδο βαθμολόγησης πολύ συγγενική με το TF-IDF, για να κατατάξει τα έγγραφα με βάση το πόσο καλά ταιριάζουν με το query. Μόλις αναγνωριστούν τα σχετικά έγγραφα, ο searcher τα ανακτά σε σειρά κατάταξης, επιστρέφοντας τα top-scoring αποτελέσματα.

Στη συνέχεια εξηγούμε το BM25.


**BM25 (Best Match 25)**

Το BM25 είναι μια προηγμένη ranking function που βελτιώνει το TF-IDF αντιμετωπίζοντας δύο βασικούς περιορισμούς:
1. Term frequency saturation (φθίνουσες αποδόσεις)
2. Document length normalization

BM25 score για ένα έγγραφο D και query Q:
$$
\text{score}(D,Q) = \sum_{i=1}^n \text{IDF}(q_i) \cdot \frac{f(q_i,D) \cdot (k_1 + 1)}{f(q_i,D) + k_1 \cdot (1 - b + b \cdot \frac{|D|}{\text{avgdl}})}
$$

Όπου:
- $f(q_i,D)$: Term frequency της λέξης qi στο έγγραφο D
- $|D|$: Μήκος του εγγράφου D
- avgdl: Μέσο μήκος εγγράφου
- $k_1$: Παράμετρος term frequency saturation (τυπικά 1.2-2.0)
- $b$: Παράμετρος length normalization (τυπικά 0.75)

**Βασικές Βελτιώσεις σε σχέση με TF-IDF**

1. Term Frequency Saturation
- TF-IDF: Γραμμική αύξηση με term frequency
- BM25: Φθίνουσες αποδόσεις καθώς αυξάνεται το term frequency
- Παράδειγμα: Εάν το "neural" εμφανίζεται 10 φορές vs 100 φορές, η διαφορά στο score δεν είναι 10x

2. Document Length Normalization
- TF-IDF: Δεν λαμβάνει υπόψη το μήκος εγγράφου
- BM25: Κανονικοποιεί με βάση το μήκος εγγράφου
- Τα μακρύτερα έγγραφα δεν παίρνουν αυτόματα υψηλότερα scores

3. Απλό Παράδειγμα
Query: "neural networks"
- Σύντομο paper με 2 αναφορές: μπορεί να πάρει υψηλότερο score
- Μακρύ paper με 3 αναφορές: μπορεί να πάρει χαμηλότερο score
- Το BM25 λαμβάνει υπόψη:
  - Μήκος εγγράφου vs. μέσο μήκος
  - Φθίνοντα value επαναλαμβανόμενων όρων
  - Σπανιότητα όρων στη συλλογή

Αυτός είναι ο λόγος που το BM25 συχνά ξεπερνά απλούστερες μεθόδους ranking.

````python
# List of book titles and their corresponding Project Gutenberg IDs
books = {
    "Roughing It": 3177,
    "Life on the Mississippi": 245,
    "The Adventures of Tom Sawyer": 74,
    "Adventures of Huckleberry Finn": 76
}
````

````python
import gutenbergpy.textget
import re

def clean_text(text):
    """Clean up the text by removing headers, footers, chapter titles, etc."""
    # Convert bytes to string if needed
    if isinstance(text, bytes):
        text = text.decode('utf-8')

    # Remove Project Gutenberg headers and footers
    text = gutenbergpy.textget.strip_headers(text.encode('utf-8')).decode('utf-8')

    # Remove image references and file sizes
    text = re.sub(r'\([0-9]+K\)', '', text)
    text = re.sub(r'\.jpg', '', text)

    # Remove table of contents and metadata
    text = re.sub(r'^.*?CONTENTS.*?\n.*?\n', '', text, flags=re.DOTALL)
    text = re.sub(r'^.*?TABLE OF CONTENTS.*?\n.*?\n', '', text, flags=re.DOTALL)
    text = re.sub(r'^.*?By.*?\n.*?\n', '', text, flags=re.DOTALL)
    text = re.sub(r'^.*?PREFATORY.*?\n.*?\n', '', text, flags=re.DOTALL)
    text = re.sub(r'^.*?INTRODUCTION.*?\n.*?\n', '', text, flags=re.DOTALL)

    # Remove chapter headings but keep the content
    text = re.sub(r'CHAPTER [IVXLC]+\.*\s*\n', '', text)
    text = re.sub(r'ILLUSTRATIONS?\s*\n+(?:[^\n]+\n+)*', '', text)

    # Remove extra whitespace but preserve paragraph breaks
    text = re.sub(r'[ \t]+', ' ', text)
    text = re.sub(r'\n{3,}', '\n\n', text)

    return text.strip()

def split_into_paragraphs(text, min_length=50):
    """Split text into meaningful paragraphs."""
    # Split on double newlines
    paragraphs = re.split(r'\n\s*\n', text)

    # Clean up each paragraph
    paragraphs = [p.strip() for p in paragraphs]

    # Remove short paragraphs and empty lines
    paragraphs = [p for p in paragraphs if len(p) >= min_length]

    # Remove paragraphs that are just lists or titles
    paragraphs = [p for p in paragraphs if not re.match(r'^[A-Z][a-z]*\s*$', p)]
    paragraphs = [p for p in paragraphs if not re.match(r'^[IVXLC]+\.', p)]

    return paragraphs

def download_and_process_books(books):
    corpus = {}
    doc_id = 0

    for title, book_id in books.items():
        print(f"\nProcessing {title}...")

        # Download the book
        raw_book = gutenbergpy.textget.get_text_by_id(book_id)

        # Clean the text
        clean_book = clean_text(raw_book)
        print(f"Text length after cleaning: {len(clean_book)}")

        # Split into paragraphs
        paragraphs = split_into_paragraphs(clean_book)

        # Add to corpus with metadata
        for para in paragraphs:
            corpus[f"doc_{doc_id}"] = {
                "text": para,
                "book": title,
                "book_id": book_id
            }
            doc_id += 1

        print(f"Added {len(paragraphs)} paragraphs from {title}")

    return corpus

# Orocess the corpus
mt_corpus = download_and_process_books(books)

````
