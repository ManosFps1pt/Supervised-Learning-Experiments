# Εισαγωγή στα Word Embeddings και Word2Vec

Στη σύγχρονη επεξεργασία φυσικής γλώσσας (NLP), τα embeddings λέξεων (word embeddings) είναι ένας τρόπος να μετατρέπουμε τις λέξεις σε διανύσματα (vectors) πραγματικών αριθμών, ώστε να μπορούν να χρησιμοποιηθούν από αλγορίθμους μηχανικής μάθησης. Τα embeddings αυτά συλλαμβάνουν σημασιολογικές και συντακτικές σχέσεις μεταξύ των λέξεων.

Το Word2Vec είναι ένα από τα πρώτα και πιο γνωστά μοντέλα για τη δημιουργία τέτοιων embeddings. Εκπαιδεύεται ώστε λέξεις με παρόμοιο νόημα να έχουν κοντινά διανύσματα στον χώρο.

Τα μεγάλα γλωσσικά μοντέλα (LLMs) όπως το BERT και το GPT υπολογίζουν embeddings όχι μόνο για λέξεις, αλλά και για ολόκληρα κείμενα (text embeddings), λαμβάνοντας υπόψη το συμφραζόμενο (context). Τα embeddings που παράγονται από LLMs είναι πολύ πιο πλούσια και δυναμικά σε σχέση με τα στατικά word embeddings του Word2Vec. Αυτά θα έχουμε την ευκαιρία να τα μελετήσουμε στις επόμενες διαλέξεις.

## Εγκατάσταση της βιβλιοθήκης gensim

Η βιβλιοθήκη `gensim` μας επιτρέπει να φορτώσουμε και να δουλέψουμε με προεκπαιδευμένα μοντέλα Word2Vec.

````python
!pip install -q gensim gdown
````

````python
import gensim
import gdown

# Google Drive file
file_id = '1u4vOq5nr2LFR3pCYq2g_0sldYVW3jdV6'
output_path = '/tmp/GoogleNews-vectors-negative300.bin'

# Download the file using gdown
gdown.download(id=file_id, output=output_path, quiet=False)

# Define the path to your Word2Vec model
word2vec_path = output_path

# Load the model (binary format must be True)
model = gensim.models.KeyedVectors.load_word2vec_format(word2vec_path, binary=True)
````

## Ενσωματώσεις λέξεων

Μπορούμε να χρησιμοποιήσουμε τώρα το μοντέλο. Κάθε λέξη στο λεξικό του Word2Vec, έχει ένα embedding σε 300 διαστάσεις:
$$
{\rm Word2Vec}: \{{\rm Λεξιλόγιο}\} \longrightarrow \mathbb{R}^{300}
$$

Οπότε, για οποιαδήποτε λέξει, π.χ. "king", ``model["king"]`` είναι διάνυσμα σε 300 διαστάσεις.



````python
# Test word embedding
vector = model["king"]  # Get embedding for "king"
print("Vector size:", len(vector))
````

````python
print(vector)
````

## Σημασιολογική Ομοιότητα

Το Word2Vec μοντέλο μας προσφέρει κάποιες χρήσιμες συναρτήσεις που μπορούμε εύκολα να χρησιμοποιήσουμε.

````python
# Find most similar words to "Parthenon"
for word in model.most_similar("Parthenon"):
  print(word)
````

````python
for word in model.most_similar("horse"):
  print(word)
````

````python
for word in model.most_similar("snow"):
  print(word)
````

## Άλλα Παραδείγματα

````python
# Compute cosine similarity
similarity = model.similarity("dog", "cat")
print("Cosine Similarity (Word2Vec):", similarity)
````

## Εσωτερικό Γινόμενο

Αυτή η ομοιότητα είναι ακριβώς το εσωτερικό γινόμενο!

````python
import numpy as np
dog_embedding = model["dog"]
cat_embedding = model["cat"]

dog_embedding_n = dog_embedding / np.linalg.norm(dog_embedding)
cat_embedding_n = cat_embedding / np.linalg.norm(cat_embedding)

# compute similarity / inner product
similarity = np.dot(dog_embedding_n, cat_embedding_n)

print("Cosine Similarity (Word2Vec):", similarity)
````

### Σημασιολογική Ομοιότητα

Όπως έχουμε δει, οι ενσωματώσεις επιχειρούν να αντικαταστίσουν τη σημασιολογική ομοιότητα με την Ευκλείδια απόσταση.

Μπορούμε μόνοι μας να συγκρίνουμε την ομοιότητα που κρίνει το μοντέλο. Ας δοκιμάσουμε τις τρείς λέξεις που αναφέραμε στις διαφάνειες:
* σκύλος (dog)
* γάτα (cat)
* παγωτό (ice cream)

````python
dog_embedding_n = model["dog"]/np.linalg.norm(model["dog"])
cat_embedding_n = model["cat"]/np.linalg.norm(model["cat"])
ice_cream_embedding_n = model["ice_cream"]/np.linalg.norm(model["ice_cream"])
````

````python
print(np.dot(dog_embedding_n, cat_embedding_n))
print(np.dot(dog_embedding_n, ice_cream_embedding_n))
print(np.dot(cat_embedding_n, ice_cream_embedding_n))
````
