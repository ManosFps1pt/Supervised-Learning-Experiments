```python
def predict_positions_for_test(bodies_df, train_df, test_df, parent_submission_df):
    # Subtask 2 baseline: return the last known position of each body.
    train_sorted = train_df.sort_values(['system_id', 'body_id', 'time'])
    last_pos = (
        train_sorted
        .groupby(['system_id', 'body_id'])[['x', 'y']]
        .last()
        .reset_index()
    )
    out = test_df.merge(last_pos, on=['system_id', 'body_id'], how='left')
    out[['x', 'y']] = out[['x', 'y']].fillna(0.0)
    return out[['system_id', 'time', 'body_id', 'x', 'y']]
```
