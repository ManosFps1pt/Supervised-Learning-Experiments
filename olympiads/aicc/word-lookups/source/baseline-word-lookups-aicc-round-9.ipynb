```python
import pandas as pd
import ast
```

```python
train_df = pd.read_csv("/kaggle/input/competitions/aicc-round-9-word-lookups/train.csv")
test_df = pd.read_csv("/kaggle/input/competitions/aicc-round-9-word-lookups/test.csv")

print("Train:")
display(train_df.head())

print("Test:")
display(test_df.head())
```

```python
print(train_df.columns)
print(test_df.columns)

print("\nExample chars:")
print(test_df.iloc[0]["chars"])
```

```python
predictions = []

for chars in test_df["chars"]:
    chars = ast.literal_eval(chars)
    
    tags = ["S"] * len(chars)
    
    predictions.append(str(tags))

print(predictions[0])
```

```python
submission = pd.DataFrame({
    "id": test_df["id"],
    "bio_tags": predictions
})

submission.head()
```

```python
submission.to_csv("submission.csv", index=False)

print("Saved submission.csv")
```
