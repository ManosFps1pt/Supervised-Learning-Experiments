```python
import pandas as pd
import ast
from pathlib import Path
```

```python
DATA_BASE = Path.cwd().parent / "data"
train_df = pd.read_csv(DATA_BASE / "train.csv")
test_df = pd.read_csv(DATA_BASE / "test.csv")

print("Train:")
display(train_df.head())

print("Test:")
display(test_df.head())
```

```python
print(train_df.columns)
print(test_df.columns)

print("\nExample chars:")
print(test_df.iloc[0]["chars"])
```

```python
predictions = []

for chars in test_df["chars"]:
    chars = ast.literal_eval(chars)
    
    tags = ["S"] * len(chars)
    
    predictions.append(str(tags))

print(predictions[0])
```

```python
submission = pd.DataFrame({
    "id": test_df["id"],
    "bio_tags": predictions
})

submission.head()
```

```python
submission.to_csv(Path.cwd().parent / "submission.csv", index=False)

print("Saved submission.csv")
```
