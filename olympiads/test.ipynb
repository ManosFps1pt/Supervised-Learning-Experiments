```python
from inspect import signature

def print_dir(obj):
    print(f"--- |{obj.__class__}| ---")
    for i in dir(obj):
        if i.startswith("_"):
            continue
        print(i)

def probe_callable(fn):
    print("CALLABLE:", fn)
    print(signature(fn))
    print(fn.__doc__)
```

```python
import numpy as np
print_dir(np)
```

```python
probe_callable(np.array)
```

```python
help(np.ndarray)
```

```python
from sklearn.linear_model import LogisticRegression

probe_callable(LogisticRegression)
```

```python
from transformers import AutoModel, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
probe_callable(tokenizer.__call__)
```
